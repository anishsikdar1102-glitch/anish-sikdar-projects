```python
# 🏠 House Price Prediction using Multiple & Polynomial Linear Regression

# **Machine Learning Project**
# **Under the Guidance of Pavinder Yadav**
# **Dataset:** Limited house-price dataset (`Housing.csv`)

# This Google Colab notebook contains **two separate regression pipelines**:

# 1. **Multiple Linear Regression (MLR)**
# 2. **Polynomial Linear Regression (PLR)**

# **Workflow:** Import data → Basic preprocessing → Skip feature management → Visualization → Train/test split → Model training → Coefficients & intercept → MAE/MSE/RMSE/R² → Visualization → New-house prediction.

# The notebook has been cleaned to contain **only MLR and PLR**, with no Simple Linear Regression section or duplicate pipeline.
# ## 1. Upload and import the dataset
import io
import pandas as pd
from google.colab import files

uploaded = files.upload()
file_name = next(iter(uploaded))

df = pd.read_csv(io.BytesIO(uploaded[file_name]))
print("Dataset shape:", df.shape)
display(df.head())
df.info()
## 2. Basic data preprocessing (feature management skipped)
# Basic preprocessing only
print("Missing values before preprocessing:")
print(df.isnull().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Fill numeric missing values with the median
numeric_cols = df.select_dtypes(include="number").columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# Fill categorical missing values with the mode
categorical_cols = df.select_dtypes(exclude="number").columns
for col in categorical_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing values after preprocessing:")
print(df.isnull().sum())

# Feature management is intentionally skipped.
# We only prepare usable predictors for Multiple Linear Regression.
## 3. Define `x_MLR` and `y_MLR`
# Automatically identify the target column
target_col = "price" if "price" in df.columns else df.select_dtypes(include="number").columns[-1]

# Use all remaining columns as predictors
x_MLR = df.drop(columns=[target_col])
y_MLR = df[target_col]

# Convert categorical predictors to dummy variables so sklearn can train the model
x_MLR = pd.get_dummies(x_MLR, drop_first=True, dtype=float)

print("Target column:", target_col)
print("x_MLR shape:", x_MLR.shape)
print("y_MLR shape:", y_MLR.shape)
display(x_MLR.head())
## 4. Visualize the data using a scatter plot
import matplotlib.pyplot as plt

# Plot the first numeric predictor against house price
numeric_predictors = x_MLR.select_dtypes(include="number").columns

if len(numeric_predictors) > 0:
    plot_col = "area" if "area" in x_MLR.columns else numeric_predictors[0]
    plt.figure(figsize=(8, 5))
    plt.scatter(x_MLR[plot_col], y_MLR, alpha=0.7)
    plt.xlabel(plot_col)
    plt.ylabel(target_col)
    plt.title(f"{plot_col} vs {target_col}")
    plt.grid(True)
    plt.show()
else:
    print("No numeric predictor is available for a scatter plot.")
## 5. Split the data into training and testing sets
from sklearn.model_selection import train_test_split

x_MLR_train, x_MLR_test, y_MLR_train, y_MLR_test = train_test_split(
    x_MLR, y_MLR, test_size=0.20, random_state=42
)

print("Training features:", x_MLR_train.shape)
print("Testing features:", x_MLR_test.shape)
print("Training target:", y_MLR_train.shape)
print("Testing target:", y_MLR_test.shape)
## 6. Train the Multiple Linear Regression model
from sklearn.linear_model import LinearRegression

MLR_model = LinearRegression()
MLR_model.fit(x_MLR_train, y_MLR_train)

print("Multiple Linear Regression model trained successfully.")
## 7. Display slopes (coefficients) and intercept
print("Intercept (b0):", MLR_model.intercept_)

coefficients_df = pd.DataFrame({
    "Feature": x_MLR.columns,
    "Slope / Coefficient": MLR_model.coef_
})

print("\nSlope values for all predictors:")
display(coefficients_df)
## 8. Make predictions and calculate accuracy metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_MLR_pred = MLR_model.predict(x_MLR_test)

mae = mean_absolute_error(y_MLR_test, y_MLR_pred)
mse = mean_squared_error(y_MLR_test, y_MLR_pred)
rmse = mse ** 0.5
r2 = r2_score(y_MLR_test, y_MLR_pred)

print("MAE  :", mae)
print("MSE  :", mse)
print("RMSE :", rmse)
print("R² Score:", r2)

metrics_df = pd.DataFrame({
    "Metric": ["MAE", "MSE", "RMSE", "R² Score"],
    "Value": [mae, mse, rmse, r2]
})
display(metrics_df)
## 9. Visualize actual vs predicted house prices
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(y_MLR_test, y_MLR_pred, alpha=0.7)

min_val = min(y_MLR_test.min(), y_MLR_pred.min())
max_val = max(y_MLR_test.max(), y_MLR_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], "r--", label="Perfect prediction")

plt.xlabel("Actual House Price")
plt.ylabel("Predicted House Price")
plt.title("Multiple Linear Regression: Actual vs Predicted")
plt.legend()
plt.grid(True)
plt.show()
## 10. Regression-line style visualization for one predictor
# For MLR, one single regression line cannot represent all predictors in one 2D plot.
# This visualization compares actual and model-predicted prices against one selected predictor.
if len(numeric_predictors) > 0:
    plot_col = "area" if "area" in x_MLR.columns else numeric_predictors[0]

    temp = pd.DataFrame({
        plot_col: x_MLR_test[plot_col],
        "Actual Price": y_MLR_test,
        "Predicted Price": y_MLR_pred
    }).sort_values(plot_col)

    plt.figure(figsize=(9, 5))
    plt.scatter(temp[plot_col], temp["Actual Price"], alpha=0.6, label="Actual data")
    plt.plot(temp[plot_col], temp["Predicted Price"], linewidth=2, label="MLR predicted values")
    plt.xlabel(plot_col)
    plt.ylabel(target_col)
    plt.title(f"MLR Prediction Trend using {plot_col}")
    plt.legend()
    plt.grid(True)
    plt.show()
## 11. Predict the price of a new house
# Example: create a new observation from an existing row.
# Replace the values below with your own house details when needed.

new_house = x_MLR.iloc[[0]].copy()

# Example changes only if these columns exist
if "area" in new_house.columns:
    new_house["area"] = 3000
if "bedrooms" in new_house.columns:
    new_house["bedrooms"] = 3
if "bathrooms" in new_house.columns:
    new_house["bathrooms"] = 2

predicted_price = MLR_model.predict(new_house)

print("Predicted house price:", predicted_price[0])
## Conclusion

# The experiment successfully trains a **Multiple Linear Regression** model for house-price prediction using `x_MLR` and `y_MLR`. It includes basic preprocessing, visualization, train/test splitting, model training, coefficients and intercept, accuracy metrics, prediction visualization, and new-house prediction.
# PART 2 — Polynomial Linear Regression (PLR)

# This is a **separate pipeline** from the MLR pipeline.

# The model uses a **degree-5 polynomial** of `area`:

# **Price = b₀ + b₁×Area + b₂×Area² + b₃×Area³ + b₄×Area⁴ + b₅×Area⁵**

# Required variable names:

# - `X_PLR`
# - `Y_PLR`
## 1. Import libraries and data preprocessing (basic)
# Imports for the separate Polynomial Linear Regression pipeline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Use the dataset loaded at the beginning of the notebook.
# Basic preprocessing only: select area and price, then remove missing rows.
data_PLR = df[['area', 'price']].copy()

print("Missing values before preprocessing:")
print(data_PLR.isnull().sum())

data_PLR = data_PLR.dropna()

print("\nMissing values after preprocessing:")
print(data_PLR.isnull().sum())

print("\nShape after preprocessing:", data_PLR.shape)
display(data_PLR.head())
## 2. Create `X_PLR` and `Y_PLR`
# X_PLR = input feature (area)
# Y_PLR = target (house price)

# Define data_PLR within this cell to resolve NameError
data_PLR = df[['area', 'price']].copy()
data_PLR = data_PLR.dropna()

X_PLR = data_PLR[['area']]
Y_PLR = data_PLR['price']

print("X_PLR shape:", X_PLR.shape)
print("Y_PLR shape:", Y_PLR.shape)
## 3. Visualize the original data with a scatter plot
plt.figure(figsize=(9, 6))
plt.scatter(X_PLR['area'], Y_PLR, alpha=0.6)
plt.xlabel('Area (sq. ft.)')
plt.ylabel('House Price')
plt.title('House Area vs House Price')
plt.grid(True)
plt.show()
## 4. Split the data into training and testing sets
X_train_PLR, X_test_PLR, Y_train_PLR, Y_test_PLR = train_test_split(
    X_PLR,
    Y_PLR,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train_PLR))
print("Testing samples :", len(X_test_PLR))
## 5. Convert the input into polynomial features
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split # Added import

# Degree of polynomial
degree_PLR = 5


poly_PLR = PolynomialFeatures(
    degree=degree_PLR,
    include_bias=False
)

# Define X_train_PLR and X_test_PLR within this cell
X_train_PLR, X_test_PLR, Y_train_PLR, Y_test_PLR = train_test_split(
    X_PLR,
    Y_PLR,
    test_size=0.20,
    random_state=42
)

X_train_poly_PLR = poly_PLR.fit_transform(X_train_PLR)
X_test_poly_PLR = poly_PLR.transform(X_test_PLR)

print("Polynomial feature names:")
print(poly_PLR.get_feature_names_out(['area']))

print("\nTraining polynomial data shape:", X_train_poly_PLR.shape)
print("Testing polynomial data shape :", X_test_poly_PLR.shape)
## 6. Train the Polynomial Linear Regression model
model_PLR = LinearRegression()

model_PLR.fit(
    X_train_poly_PLR,
    Y_train_PLR
)

print("Polynomial Linear Regression model trained successfully.")
## 7. Get polynomial coefficients and intercept
coefficients_PLR = model_PLR.coef_
intercept_PLR = model_PLR.intercept_
feature_names_PLR = poly_PLR.get_feature_names_out(['area'])

print(f"Intercept: {intercept_PLR:.6f}")

for feature_name, coefficient in zip(feature_names_PLR, coefficients_PLR):
    print(f"Coefficient for {feature_name}: {coefficient:.6f}")

print("\nPolynomial regression equation:")
equation_terms = [f"({coef:.6f} × {name})" for name, coef in zip(feature_names_PLR, coefficients_PLR)]
print("Price = " + " + ".join(equation_terms) + f" + {intercept_PLR:.6f}")
## 8. Accuracy / evaluation metrics
Y_pred_PLR = model_PLR.predict(X_test_poly_PLR)

mae_PLR = mean_absolute_error(Y_test_PLR, Y_pred_PLR)
mse_PLR = mean_squared_error(Y_test_PLR, Y_pred_PLR)
rmse_PLR = np.sqrt(mse_PLR)
r2_PLR = r2_score(Y_test_PLR, Y_pred_PLR)

print("Polynomial Linear Regression Evaluation")
print("---------------------------------------")
print(f"R² Score : {r2_PLR:.6f}")
print(f"MAE      : {mae_PLR:,.2f}")
print(f"MSE      : {mse_PLR:,.2f}")
print(f"RMSE     : {rmse_PLR:,.2f}")
## 9. Visualize the polynomial regression curve
plt.figure(figsize=(9, 6))

# Actual observations
plt.scatter(
    X_PLR['area'],
    Y_PLR,
    alpha=0.5,
    label='Actual data'
)

# Smooth polynomial curve
X_curve_PLR = np.linspace(
    X_PLR['area'].min(),
    X_PLR['area'].max(),
    300
).reshape(-1, 1)

X_curve_poly_PLR = poly_PLR.transform(X_curve_PLR)
Y_curve_PLR = model_PLR.predict(X_curve_poly_PLR)

plt.plot(
    X_curve_PLR,
    Y_curve_PLR,
    linewidth=2,
    label=f'Polynomial curve (degree {degree_PLR})'
)

plt.xlabel('Area (sq. ft.)')
plt.ylabel('House Price')
plt.title('Polynomial Linear Regression: Area vs House Price')
plt.legend()
plt.grid(True)
plt.show()
## 10. Predict a house price using `X_PLR`
# Enter a new house area in square feet
new_area_PLR = 5000

# Create prediction input using the requested X_PLR naming convention
X_new_PLR = pd.DataFrame({'area': [new_area_PLR]})

# Convert the new input using the SAME polynomial transformer
X_new_poly_PLR = poly_PLR.transform(X_new_PLR)

predicted_price_PLR = model_PLR.predict(X_new_poly_PLR)[0]

print(f"Area: {new_area_PLR:,} sq. ft.")
print(f"Predicted house price: {predicted_price_PLR:,.2f}")
## Final Summary

# You now have **two separate regression pipelines**:

# | Pipeline | Input | Target | Model |
# |--|---|---|---|
# | **Multiple Linear Regression (MLR)** | `x_MLR` | `y_MLR` | `LinearRegression()` with multiple predictors |
# | **Polynomial Linear Regression (PLR)** | `X_PLR` | `Y_PLR` | `PolynomialFeatures` + `LinearRegression()` |

### Important
# - The MLR pipeline uses all available predictors after basic preprocessing; feature selection/management is intentionally skipped.
# - The PLR pipeline models `area` against `price`.
# - The PLR polynomial degree is **5**, matching the actual code.
# - Both pipelines report **MAE, MSE, RMSE, and R²** and include prediction code.
## Conclusion

# This project demonstrates **Multiple Linear Regression and Polynomial Linear Regression** for house-price prediction using a limited dataset. Both pipelines include data import, basic preprocessing, visualization, train/test splitting, model training, coefficients/intercept, evaluation metrics, and new-house prediction.

# The two models can be compared using their MAE, MSE, RMSE, and R² results after running all cells in Google Colab.
```
