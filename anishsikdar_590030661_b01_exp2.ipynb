```python
# ============================================================
# IRIS FLOWER CLASSIFICATION - MULTIPLE MACHINE LEARNING MODELS
# ============================================================


# ============================================================
# STEP 1: IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    classification_report
)


# ============================================================
# STEP 2: LOAD THE DATASET
# ============================================================

# Upload the CSV file in Google Colab first.
# Then update the filename/path if necessary.

df = pd.read_csv("01_iris_flower_classification.csv")

print("First 5 rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
print(df.info())


# ============================================================
# STEP 3: DEFINE FEATURES AND TARGET
# NO PREPROCESSING
# NO FEATURE MANAGEMENT
# ============================================================

X = df.drop("species", axis=1)

Y = df["species"]

print("\nFeatures:")
display(X.head())

print("\nTarget:")
display(Y.head())


# ============================================================
# STEP 4: DIRECTLY SPLIT THE DATA
# ============================================================

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("Y_train shape:", Y_train.shape)
print("Y_test shape:", Y_test.shape)


# ============================================================
# STEP 5: CREATE MULTIPLE MACHINE LEARNING MODELS
# ============================================================

models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),

    "SVM": SVC(
        kernel="linear"
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    )
}


# ============================================================
# STEP 6: LOOP THROUGH ALL MODELS
# TRAIN -> PREDICT -> ACCURACY -> CONFUSION MATRIX
# ============================================================

accuracy_results = {}

for model_name, model in models.items():

    print("\n")
    print("=" * 60)
    print("MODEL:", model_name)
    print("=" * 60)

    # Train the model
    model.fit(X_train, Y_train)

    # Generate predictions
    Y_pred = model.predict(X_test)

    # Calculate accuracy
    accuracy = accuracy_score(Y_test, Y_pred)

    accuracy_results[model_name] = accuracy

    print("\nPredictions:")
    print(Y_pred)

    print("\nAccuracy:")
    print(round(accuracy * 100, 2), "%")

    print("\nClassification Report:")
    print(classification_report(Y_test, Y_pred))

    # Confusion Matrix
    cm = confusion_matrix(Y_test, Y_pred)

    print("\nConfusion Matrix:")
    print(cm)

    # Display Confusion Matrix
    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=model.classes_
    )

    disp.plot()

    plt.title(
        f"Confusion Matrix - {model_name}"
    )

    plt.show()


# ============================================================
# STEP 7: DISPLAY ACCURACY OF ALL MODELS
# ============================================================

print("\n")
print("=" * 60)
print("MODEL ACCURACY COMPARISON")
print("=" * 60)

for model_name, accuracy in accuracy_results.items():

    print(
        f"{model_name}: "
        f"{round(accuracy * 100, 2)}%"
    )


# ============================================================
# STEP 8: LOGISTIC REGRESSION VISUALIZATION
# ============================================================

# For visualization, we use two features only:
#
# X-axis -> sepal_length_cm
# Y-axis -> sepal_width_cm
#
# We also use two classes:
#
# Iris-setosa
# Iris-versicolor
#
# This makes it possible to visualize a straight
# logistic regression classification boundary.


binary_df = df[
    df["species"].isin(
        ["Iris-setosa", "Iris-versicolor"]
    )
].copy()


# Select two features for 2D visualization

X_logistic = binary_df[
    [
        "sepal_length_cm",
        "sepal_width_cm"
    ]
]

Y_logistic = binary_df["species"]


# ============================================================
# STEP 9: SPLIT DATA FOR LOGISTIC REGRESSION VISUALIZATION
# ============================================================

X_train_logistic, X_test_logistic, \
Y_train_logistic, Y_test_logistic = train_test_split(

    X_logistic,
    Y_logistic,

    test_size=0.2,

    random_state=42,

    stratify=Y_logistic
)


# ============================================================
# STEP 10: TRAIN LOGISTIC REGRESSION
# ============================================================

logistic_model = LogisticRegression()

logistic_model.fit(
    X_train_logistic,
    Y_train_logistic
)


# ============================================================
# STEP 11: GENERATE LOGISTIC REGRESSION PREDICTIONS
# ============================================================

Y_pred_logistic = logistic_model.predict(
    X_test_logistic
)


print("\n")
print("=" * 60)
print("LOGISTIC REGRESSION RESULTS")
print("=" * 60)

print("\nPredictions:")
print(Y_pred_logistic)


# ============================================================
# STEP 12: LOGISTIC REGRESSION ACCURACY
# ============================================================

logistic_accuracy = accuracy_score(
    Y_test_logistic,
    Y_pred_logistic
)

print("\nAccuracy:")
print(
    round(logistic_accuracy * 100, 2),
    "%"
)


# ============================================================
# STEP 13: LOGISTIC REGRESSION CONFUSION MATRIX
# ============================================================

cm_logistic = confusion_matrix(
    Y_test_logistic,
    Y_pred_logistic
)

print("\nConfusion Matrix:")
print(cm_logistic)


disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_logistic,
    display_labels=logistic_model.classes_
)

disp.plot()

plt.title(
    "Confusion Matrix - Logistic Regression"
)

plt.show()


# ============================================================
# STEP 14: SCATTER PLOT OF THE DATA
# ============================================================

plt.figure(figsize=(8, 6))

for species in binary_df["species"].unique():

    subset = binary_df[
        binary_df["species"] == species
    ]

    plt.scatter(

        subset["sepal_length_cm"],

        subset["sepal_width_cm"],

        label=species
    )


plt.xlabel("Sepal Length (cm)")

plt.ylabel("Sepal Width (cm)")

plt.title(
    "Iris Dataset Scatter Plot"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# STEP 15: GET LOGISTIC REGRESSION COEFFICIENTS
# ============================================================

# Logistic Regression equation:
#
# b0 + b1*x1 + b2*x2 = 0
#
# Classification boundary:
#
# x2 = -(b0 + b1*x1) / b2


b1 = logistic_model.coef_[0][0]

b2 = logistic_model.coef_[0][1]

b0 = logistic_model.intercept_[0]


print("\n")
print("=" * 60)
print("LOGISTIC REGRESSION COEFFICIENTS")
print("=" * 60)

print("Coefficient for Sepal Length:", b1)

print("Coefficient for Sepal Width:", b2)

print("Intercept:", b0)


# ============================================================
# STEP 16: CALCULATE SLOPE AND INTERCEPT
# ============================================================

# From:
#
# b0 + b1*x + b2*y = 0
#
# Rearranging:
#
# y = -(b0/b2) - (b1/b2)*x
#
# Therefore:
#
# Slope = -b1 / b2
#
# Line Intercept = -b0 / b2


slope = -b1 / b2

line_intercept = -b0 / b2


print("\nClassification Line Slope:")
print(slope)

print("\nClassification Line Intercept:")
print(line_intercept)


# ============================================================
# STEP 17: PLOT DATA + CLASSIFICATION LINE
# ============================================================

plt.figure(figsize=(10, 7))


# Scatter Plot

for species in binary_df["species"].unique():

    subset = binary_df[
        binary_df["species"] == species
    ]

    plt.scatter(

        subset["sepal_length_cm"],

        subset["sepal_width_cm"],

        label=species
    )


# Generate X values for the classification line

x_values = np.linspace(

    X_logistic.iloc[:, 0].min() - 0.5,

    X_logistic.iloc[:, 0].max() + 0.5,

    100
)


# Calculate Y values

y_values = (

    slope * x_values

    +

    line_intercept
)


# Plot Classification Line

plt.plot(

    x_values,

    y_values,

    linewidth=2,

    label="Logistic Regression Classification Boundary"
)


plt.xlabel("Sepal Length (cm)")

plt.ylabel("Sepal Width (cm)")

plt.title(
    "Logistic Regression Classification Boundary"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# STEP 18: FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("FINAL SUMMARY")
print("=" * 60)

print("\nModels Trained:")

for model_name in models.keys():

    print("-", model_name)


print("\nAccuracy Results:")

for model_name, accuracy in accuracy_results.items():

    print(
        f"{model_name}: "
        f"{round(accuracy * 100, 2)}%"
    )


print("\nLogistic Regression Classification Line:")

print(
    "y =",
    slope,
    "* x +",
    line_intercept
)
```
